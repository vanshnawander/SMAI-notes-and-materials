$$
f'(x) \approx \frac{f(x+h) - f(x)}{h}
$$

$$
f'(x) \approx \frac{f(x) - f(x-h)}{h}
$$

$$
f'(x) \approx \frac{f(x+h) - f(x-h)}{2h}
$$

```python
import numpy as np


def numerical_derivative(f, x, h=1e-5, method="central"):
    """
    Compute numerical derivative of a scalar function f at point x.

    method: 'forward', 'backward', or 'central'
    """
    if method == "forward":
        return (f(x + h) - f(x)) / h
    elif method == "backward":
        return (f(x) - f(x - h)) / h
    elif method == "central":
        return (f(x + h) - f(x - h)) / (2 * h)
    else:
        raise ValueError("method must be one of: 'forward', 'backward', 'central'")


def approximate_gradient_1d(f, x, h=1e-7):
    """For 1D functions, gradient is the derivative."""
    return numerical_derivative(f, x, h=h, method="central")
```

$$
f(x) = \sin(x^2 + 3x)
$$

```python
# f(x) = sin(x^2 + 3x)
def f1(x):
    return np.sin(x**2 + 3*x)

x1 = 1.0
h = 1e-5

print(f"Function 1: f(x) = sin(x^2 + 3x)")
print(f"x = {x1}")
print(f"Forward difference  : {numerical_derivative(f1, x1, h, method='forward')}")
print(f"Backward difference : {numerical_derivative(f1, x1, h, method='backward')}")
print(f"Central difference  : {numerical_derivative(f1, x1, h, method='central')}")
print(f"Approx gradient     : {approximate_gradient_1d(f1, x1, h)}")
```

$$
f(x) = \frac{\log(x^2 + 1)}{x}
$$

```python
# f(x) = log(x^2 + 1) / x
def f2(x):
    return np.log(x**2 + 1) / x

x2 = 2.0  # choose non-zero x because of division by x
h = 1e-5

print(f"Function 2: f(x) = log(x^2 + 1) / x")
print(f"x = {x2}")
print(f"Forward difference  : {numerical_derivative(f2, x2, h, method='forward')}")
print(f"Backward difference : {numerical_derivative(f2, x2, h, method='backward')}")
print(f"Central difference  : {numerical_derivative(f2, x2, h, method='central')}")
print(f"Approx gradient     : {approximate_gradient_1d(f2, x2, h)}")
```

$$
f(x) = \log(1 + e^{x^2})
$$

```python
# f(x) = log(1 + e^(x^2))
def f3(x):
    return np.log(1 + np.exp(x**2))

x3 = 1.5
h = 1e-5

print(f"Function 3: f(x) = log(1 + e^(x^2))")
print(f"x = {x3}")
print(f"Forward difference  : {numerical_derivative(f3, x3, h, method='forward')}")
print(f"Backward difference : {numerical_derivative(f3, x3, h, method='backward')}")
print(f"Central difference  : {numerical_derivative(f3, x3, h, method='central')}")
print(f"Approx gradient     : {approximate_gradient_1d(f3, x3, h)}")
```
