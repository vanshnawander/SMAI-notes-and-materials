```python
from ddgs import DDGS
```

```python
import os 
currency_notes = ["1","2","5","10","20","50","100","200","500","2000"]
os.makedirs('dataset/train-dataset', exist_ok=True)
for i in currency_notes:
    os.makedirs(f'dataset/train-dataset/{i}', exist_ok=True)
```

```python
import urllib3 
import shutil
failures = []
def save_images(results, currenct_index):
    http = urllib3.PoolManager()
    for i in range(len(results)):
        try:
            url = results[i]['image']
            filename = f'dataset/train-dataset/{currency_notes[currenct_index]}/{currency_notes[currenct_index]}note_{i}.jpg'
            with http.request('GET', url, preload_content=False) as r:
                with open(filename, 'wb') as out_file:
                    shutil.copyfileobj(r, out_file)
        except Exception as e:
            print(f"Error downloading image {i} for currency {currency_notes[currenct_index]}: {e}")
            print(f"URL: {url}")
            failures.append((currency_notes[currenct_index], i, url))
```

```python
import time
for i in range(len(currency_notes)):
    try:
        results = DDGS().images(f"{currency_notes[i]} notes indian currency", max_results=50)
        time.sleep(1)
        results2 = DDGS().images(f"{currency_notes[i]} rupees notes indian currency", max_results=50)
        time.sleep(1)
        results3 = DDGS().images(f"indian currency {currency_notes[i]} rupee notes", max_results=50)
        time.sleep(1)
        results4 = DDGS().images(f"indian currency {currency_notes[i]} note only", max_results=50)
        time.sleep(1)

        save_images(results+results2+results3+results4, i)
        time.sleep(50)
    except Exception as e:
        print(f"Error for {currency_notes[i]}: {e}")
```
