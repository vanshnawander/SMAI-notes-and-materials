### Question 5: Taylor Series Approximation

The first-order Taylor approximation allows us to estimate the value of a function near a known point. Complete the code to compute this approximation.


```python
def first_order_approximation(j_old, grad_old, w_old, w_new):
    """
    j_old: Value of J at w_old
    grad_old: Gradient of J at w_old
    w_old: current weight vector
    w_new: new weight vector
    """
    # s is the displacement vector (w_new - w_old)
    s = w_new - w_old

    # TODO: Compute J_approx
    # Use np.dot for the vector multiplication
    j_approx = j_old + np.dot(grad_old, s) ## Placeholder value, replace with actual calculation

    return j_approx
```

```python
w_old = np.array([2.0])   # starting point
v_prev = np.array([0.0])  # initial velocity
learning_rate = 0.1
gamma = 0.9

# Function and gradient
def J(w):
    return w**2

def grad_J(w):
    return 2* w
```

```python


# Compute values at w_old
j_old = J(w_old)
grad_old = grad_J(w_old)

# Momentum update
w_new, v_new = momentum_update(w_old, grad_old, v_prev, learning_rate, gamma)

# True value vs approximation
j_true = J(w_new)
j_approx = first_order_approximation(j_old, grad_old, w_old, w_new)

print("w_old:", w_old)
print("grad_old:", grad_old)
print("w_new:", w_new)

print("\nTrue J(w_new):", j_true)
print("First-order Approximation:", j_approx)
```
