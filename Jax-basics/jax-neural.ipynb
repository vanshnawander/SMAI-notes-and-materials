```python
class ShallowCNN(nnx.Module):
    def __init__(self):
        super(ShallowCNN, self).__init__()
        self.conv1 = nnx.Conv2d(3, 16, kernel_size=3, stride=1, padding=1)
        self.pool = nnx.MaxPool2d(kernel_size=2, stride=2, padding=0)
        self.conv2 = nnx.Conv2d(16, 32, kernel_size=3, stride=1, padding=1)
        self.pool2 = nnx.MaxPool2d(kernel_size=2, stride=2, padding=0)
        self.fc1 = nnx.Linear(32 * 8 * 8, 10)

    def forward(self, x):
        x = self.conv1(x)
        x = self.pool(nnx.relu(x))
        x = self.conv2(x)
        x = self.pool(nnx.relu(x))
        x = x.view(-1, 32 * 8 * 8)
        x = self.fc1(x)
        return x
```
